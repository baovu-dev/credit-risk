DebtRatio: 99th percentile is still 4979 (way too high for a "ratio"), because rows with originally-missing income (now imputed to the 5400 median) produce huge debt/income ratios. Using a domain-knowledge cap of 10 instead of a percentile-based one, since the percentile itself is contaminated.

```python
df['DebtRatio'] = df['DebtRatio'].clip(upper=10)
```

RevolvingUtilization: 99th percentile (1.09) looks clean and reasonable, so percentile-based capping works fine here.

```python
# RevolvingUtilization: 99th percentile (1.09) looks clean and reasonable,
# so percentile-based capping works fine here.
df['RevolvingUtilizationOfUnsecuredLines'] = df['RevolvingUtilizationOfUnsecuredLines'].clip(
    upper=df['RevolvingUtilizationOfUnsecuredLines'].quantile(0.99)
)
```
